# Export bundle — thesis integration (Phase 6 + Phase 7)

Collects every artifact the thesis update needs: result JSONs, tables,
figures (PDF+PNG), logs, preregs, and data cards produced by notebooks
05v2/23v2/24/25/25b/15v2 (Phase 6) and 26 v1–v3 / 27 (Phase 7). Glob-based,
so filename variations are caught; missing items are listed, not fatal.
Run once; `thesis_update_bundle.zip` is then used, together with the
thesis `.tex` sources, to update the thesis text.

```python
from google.colab import drive
drive.mount('/content/drive')

import zipfile, glob
from pathlib import Path

ROOT = Path('/content/drive/MyDrive/phd_thesis')
OUT = Path('/content/thesis_update_bundle.zip')

PATTERNS = [
    # ---- result JSONs (exact numbers for the text) ----
    'data/processed/phase6_*.json',
    'data/processed/phase7_*.json',
    'data/processed/sig_15_*.json',
    'data/processed/phase5_21*.json',          # nb21 reference (control values)
    'data/processed/phase5_22*.json',          # nb22 reference (D040 baseline)
    # ---- tables ----
    'reports/tables/23_*.csv',
    'reports/tables/24_*.csv',
    'reports/tables/25_*.csv',
    'reports/tables/25b_*.csv',
    'reports/tables/26_*.csv',
    'reports/tables/27_*.csv',
    'reports/tables/15_tabsig.tex',
    'reports/tables/15_significance_*.csv',
    'reports/tables/22_per_seed_raw.csv',
    'reports/tables/22_three_regime_confirmed.csv',
    'reports/tables/09_unsw_vs_cicids_geometry.csv',
    # ---- figures (both formats; thesis uses PDF, quick checks use PNG) ----
    'reports/figures/23_*.p*',
    'reports/figures/24_*.p*',
    'reports/figures/25_*.p*',
    'reports/figures/25b_*.p*',
    'reports/figures/26_*.p*',
    'reports/figures/27_*.p*',
    # ---- logs, preregs, data cards (F/D numbers + audit trail for the text) ----
    'reports/advisor_notes/findings_log.md',
    'reports/advisor_notes/decisions_log.md',
    'reports/advisor_notes/PREREG*.md',
    'reports/advisor_notes/data_card_cse2018.md',
    'reports/advisor_notes/conversion_report_cse2018.*',
    # ---- promoted src (cited in the methods text) ----
    'src/adaptation/trigger_v3.py',
    'src/monitoring/ks_monitor.py',
]

added, missing = [], []
with zipfile.ZipFile(OUT, 'w', zipfile.ZIP_DEFLATED) as z:
    for pat in PATTERNS:
        hits = sorted(glob.glob(str(ROOT / pat)))
        if not hits:
            missing.append(pat)
            continue
        for h in hits:
            rel = str(Path(h).relative_to(ROOT))
            z.write(h, arcname=rel)
            added.append(rel)

print(f'added {len(added)} files, {OUT.stat().st_size/1024/1024:.1f} MB')
for a in added:
    print('  +', a)
if missing:
    print('\nNo match for (may just be named differently):')
    for m in missing:
        print('  -', m)
```
